```python

def alpha_beta(tree, depth, alpha, beta, maximizing):
    if depth == 0:
        return tree

    if maximizing:
        value = float('-inf')
        for i, child in enumerate(tree):
            value = max(value, alpha_beta(child, depth-1, alpha, beta, False))
            alpha = max(alpha, value)

            if alpha >= beta:
                print("Pruned subtree:", tree[i+1:])
                break

    else:
        value = float('inf')
        for i, child in enumerate(tree):
            value = min(value, alpha_beta(child, depth-1, alpha, beta, True))
            beta = min(beta, value)

            if alpha >= beta:
                print("Pruned subtree:", tree[i+1:])
                break

    return value


# Game tree
tree = [[3, 5], [2, 9]]

result = alpha_beta(tree, 2, float('-inf'), float('inf'), True)

print("\nFinal value of MAX node:", result)
```
