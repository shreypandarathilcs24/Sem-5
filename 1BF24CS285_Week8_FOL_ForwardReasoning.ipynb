```python

# FOL Resolution

def resolve(c1, c2):
    for i, (p, x, n1) in enumerate(c1):
        for j, (q, y, n2) in enumerate(c2):

            if p == q and n1 != n2:
                if x == y:
                    sub = {}
                elif x.startswith("?"):
                    sub = {x: y}
                elif y.startswith("?"):
                    sub = {y: x}
                else:
                    continue

                result = list(c1[:i] + c1[i+1:] +
                              c2[:j] + c2[j+1:])

                return tuple(
                    (p, sub.get(x, x), n)
                    for p, x, n in result
                )
    return None


def main():
    # True = Negative, False = Positive

    KB = [
        (("Cat", "?x", True), ("Mammal", "?x", False)),
        (("Cat", "Tom", False),),
        (("Mammal", "Tom", True),)
    ]

    print("Resolution Steps:")

    while True:
        new = []

        for i in range(len(KB)):
            for j in range(i + 1, len(KB)):
                r = resolve(KB[i], KB[j])

                if r is not None and r not in KB and r not in new:
                    print(r)
                    new.append(r)

                    if not r:
                        print("Query Proved!")
                        return

        if not new:
            print("Query Not Proved!")
            return

        KB.extend(new)


main()
```
